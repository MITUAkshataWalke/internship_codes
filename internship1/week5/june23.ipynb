```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.array([10,23,29,34,39,45,50])
y=np.array([14,21,31,43,48,55,58])
z=np.array([56,50,44,37,26,20,16])
```

```python
plt.figure(figsize=(16,9))
plt.subplot(2,2,1)
plt.xlabel("x values")
plt.ylabel("y values")
plt.title("line chart:x vs y")
plt.plot(x,y,marker='o',mfc='red',ls="--");
plt.grid();
plt.subplot(2,2,2)
plt.xlabel("x values")
plt.ylabel("z values")
plt.title("line chart:x vs z")
plt.plot(x,z,marker="o",mfc='red',ls="--");
plt.grid();
```

```python
courses=['python','lisp','c','R','php']
students=[68,30,47,39,60]
plt.bar(courses,students,color='cyan',edgecolor='red');
plt.xlabel("courses")
plt.ylabel("students")
plt.title("marks of students in different courses")
plt.grid()
```
