```python
#q.1. Download the seeds dataset from the below link and perform the following
# operations.
# https://archive.ics.uci.edu/ml/datasets/seeds
# 1. Read the dataset and seperate the input and output columns
df = pd.read_csv(
    "seeds_dataset.txt",
    sep=r"\s+",
    header=None
)
```

```python
X=df[[0,1,2,3,4,5,6]]
Y=df[[7]]
```

```python
X
```

```python
# 2. Plot the graph of class balancing
import seaborn as sns
import matplotlib.pyplot as plt
sns.countplot(x=df[7])
plt.title("class balancing")
plt.xlabel("class")
plt.ylabel("count")
plt.show()
```

```python
# 3. Perform train-test split with 80% and 20%
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(
    X,
    Y,
    test_size=0.20,
    random_state=42
)
```

```python
# 4. Create logistic regression model
from sklearn.linear_model import LogisticRegression
model=LogisticRegression()
model.fit(X_train,Y_train)
```

```python
# 5. Find accuracy, confusion matrix and classification report
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
y_pred=model.predict(X_test)
accuracy=accuracy_score(Y_test,y_pred)
print("Accuracy:",accuracy)
cm=confusion_matrix(Y_test,y_pred)
print("\nconfusion matrix:")
print(cm)
cr=classification_report(Y_test,y_pred)
print("\nclassification report:")
print(cr)
```

```python
# 6.Perform the prediction of single value entry.
sample=[[15.26,14.84,0.8710,5.763,3.312,2.221,5.220]]
prediction=model.predict(sample)
print("predicted class:",prediction[0])
```

```python
# 2. Download the balance scale dataset from the below link and perform
# following operations it.
# https://archive.ics.uci.edu/ml/datasets/Balance+Scale
# 1. Read the dataset and seperate the input and output columns
df=pd.read_csv("balance-scale.data",header=None)
```

```python
df
```

```python
X=df[[1,2,3,4]]
Y=df[[0]]
```

```python
X
```

```python
Y
```

```python
# 2. Plot the graph of class balancing
import matplotlib.pyplot as plt
import seaborn as sns
sns.countplot(x=df[0])
plt.xlabel("class")
plt.ylabel("count")
plt.title("class balancing")
plt.show()
```

```python
# 3. Perform train-test split with 75% and 25%
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(
    X,
    Y,
    test_size=0.25,
    random_state=42
)
```

```python
# 4. Create decision tree model
from sklearn.tree import DecisionTreeClassifier
model=DecisionTreeClassifier()
model.fit(X_train,Y_train)
```

```python
# 5. Find accuracy, confusion matrix and classification report
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
y_pred=model.predict(X_test)
accuracy=accuracy_score(Y_test,y_pred)
print("accuracy:",accuracy)
cm=confusion_matrix(Y_test,y_pred)
print("confusion matrix:")
print(cm)
cr=classification_report(Y_test,y_pred)
print("classification report:")
print(cr)
```

```python
# 6.Perform the prediction of single value entry.
prediction=model.predict([[5,2,3,1]])
print("prediction:",prediction[0])
```

```python
# 7. Plot and save the tree
from sklearn.tree import plot_tree
plt.figure(figsize=(15,10))
plot_tree(model,
         feature_names=X.columns,
          class_names=model.classes_,
         filled=True)
plt.savefig("decision_tree.png")
plt.show()
```

```python
# 3. Download the balance scale dataset from the below link and perform following operations it.
# https://archive.ics.uci.edu/ml/datasets/Computer+Hardware
# 1. Read the dataset and seperate the input and output columns
df=pd.read_csv("machine.data",header=None)
df
```

```python
X=df[[2,3,4,5,6,7]]
Y=df[[8]]
```

```python
# 2. Perform train-test split with 75% and 25%
X_train,X_test,Y_train,Y_test=train_test_split(
    X,
    Y,
    test_size=0.25,
    random_state=42
)
```

```python
# 3. Create decision tree model
from sklearn.tree import DecisionTreeClassifier
model=DecisionTreeClassifier(random_state=42)
model.fit(X_train,Y_train)
```

```python
# 4. Find Mean absolute error and accuracy
y_pred=model.predict(X_test)
from sklearn.metrics import accuracy_score,mean_absolute_error
mae=mean_absolute_error(Y_test,y_pred)
acc=accuracy_score(Y_test,y_pred)
print("Mean absolute error:",mae)
print("accuracy",acc)
```

```python
# 6. Plot and save the tree.
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
plt.figure(figsize=(15,10))
plot_tree(model,
         feature_names=X.columns,
         
         filled=True)
plt.savefig("decisiontree1.png")
plt.show();
```
